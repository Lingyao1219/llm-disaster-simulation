```python
def download_images(lat, lng, location_name=None):
   """
   Download and display street view images for given coordinates from both Mapillary and OpenStreetCam
   
   Args:
       lat (float): Latitude
       lng (float): Longitude
       location_name (str, optional): Name of the location for display purposes
   """
   if location_name:
       print(f"\nDownloading images for {location_name} ({lat}, {lng})")
   else:
       print(f"\nDownloading images for coordinates: {lat}, {lng}")
   
   # Try Mapillary
   print("\nTrying Mapillary...")
   success, result = downloader.download_mapillary_image(lat, lng)
   
   if success:
       print(f"Successfully downloaded: {result}")
       try:
           display(Image(result))
       except Exception as e:
           print(f"Error displaying Mapillary image: {e}")
   else:
       print(f"Mapillary download failed: {result}")
       
   # Try OpenStreetCam
   print("\nTrying OpenStreetCam...")
   success, result = downloader.download_openstreetcam_photo(lat, lng)
   
   if success:
       print(f"Successfully downloaded: {result}")
       try:
           display(Image(result))
       except Exception as e:
           print(f"Error displaying OpenStreetCam image: {e}")
   else:
       print(f"OpenStreetCam download failed: {result}")
```

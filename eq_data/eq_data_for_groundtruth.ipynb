```python
import json
import pandas as pd
```

```python
file_path = 'ci14607652/current/products/stationlist.json'
with open(file_path, "r") as f:
    data = json.load(f)

# Extract required data
records = []
for feature in data.get("features", []):  # Iterate through each feature
    properties = feature.get("properties", {})
    geometry = feature.get("geometry", {})

    # Check if network is "DYFI"
    if properties.get("network") == "DYFI":
        latitude = geometry.get("coordinates", [None, None])[1]  # Extract latitude
        longitude = geometry.get("coordinates", [None, None])[0]  # Extract longitude
        intensity = properties.get("intensity")  # Extract intensity
        nresp = properties.get("nresp")  # Extract nresp

        # Ensure all values exist before adding to list
        if None not in (latitude, longitude, intensity, nresp):
            records.append([latitude, longitude, intensity, nresp])

# Convert to pandas DataFrame for better visualization
df = pd.DataFrame(records, columns=["Latitude", "Longitude", "Intensity", "Nresp"])
```
